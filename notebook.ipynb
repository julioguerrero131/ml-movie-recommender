Las películas se identifican por un *id*, por lo que se hace un merge con este campo, y se eliminan los campos que se repiten en ambos df, por lo que se elimina *id* y *title_y*.
Al repetirse title en ambos df, pandas automáticamente manbtiene ambos campos cambiándolos de nombre, y por ello existe *title_x* y *title_y*.

```python
df = df1.merge(df2, left_on='movie_id', right_on='id')
df = df.drop(columns=['id'])
df = df.drop(columns=['title_y'])
df = df.rename(columns={'title_x': 'title'})
df.head()
```

# Feature Engineering

Las características a usar serán las siguientes:
+ ***movie_id* y *title***: Son identificadores necesarios de las películas.
+ ***overview*, *genres* y *keywords***: Permiten responder el ¿de qué se trata la película?
+ ***cast* y *crew***: Permiten responder el ¿quién hizo la película la película?

```python
df_movies = df[['movie_id','title','overview','genres','keywords','cast','crew']]
```

```python
df_movies.isnull().sum()
```

```python
df_movies[df_movies['overview'].isna()]
```

Los registros vacíos en la columna *overview* son rellenados con strings vacíos, para mantenerlos dentro del modelo y se tomen en cuenta en las predicciones con la información de las demás columnas.

```python
df_movies['overview'] = df_movies['overview'].fillna('')
df_movies.isnull().sum()
```

Se necesita procesar las columnas que usan JSON para solo obtener la información importante de cada una en una lista en cada celda.

```python
import ast

def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name']) 
    return L 
```

```python
df_movies['genres'] = df_movies['genres'].apply(convert)
df_movies.head(2)
```

```python
df_movies['keywords'] = df_movies['keywords'].apply(convert)
df_movies.head(2)
```

Del *cast* solo se toma los tres primeros, ya que esta ordenado en elenco en orden de importancia.

```python
def convert_top(text):
    top = 3
    return [i['name'] for i in ast.literal_eval(text)[:top]]
```

```python
df_movies['cast'] = df_movies['cast'].apply(convert_top)
df_movies.head(2)
```

```python
def fetch_director(text):
    return [i['name'] for i in ast.literal_eval(text) if i['job'] == 'Director']
```

```python
df_movies['crew'] = df_movies['crew'].apply(fetch_director)
df_movies.head(2)
```

```python
def collapse(L):
    return [i.replace(" ", "") for i in L]
```

```python
df_movies['cast'] = df_movies['cast'].apply(collapse)
df_movies['crew'] = df_movies['crew'].apply(collapse)
df_movies['genres'] = df_movies['genres'].apply(collapse)
df_movies['keywords'] = df_movies['keywords'].apply(collapse)

df_movies.head(2)
```

```python
df_movies['overview'] = df_movies['overview'].apply(lambda x:x.split())
df_movies.head(2)
```

Se crea una columna *tags* que toma toda la información de cada película de las columnas procesadas, reduciendo la dimensionalidad y dejando listo un bloque de texto para la vectorización y el modelado.

```python
df_movies['tags'] = df_movies['overview'] + df_movies['genres'] + df_movies['keywords'] + df_movies['cast'] + df_movies['crew']
df_final = df_movies.drop(columns=['overview','genres','keywords','cast','crew'])
df_final['tags'] = df_final['tags'].apply(lambda x: " ".join(x))

df_final.head()
```

```python
df_final['tags'][0]
```

# Modelo

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000,stop_words='english')
vector = cv.fit_transform(df_final['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vector)
```

```python
vector
```

```python
similarity
```

```python
def recommend(movie):
    index = df_final[df_final['title'] == movie].index[0]
    distances = sorted(list(enumerate(similarity[index])),reverse=True,key = lambda x: x[1])
    for i in distances[1:6]:
        print(df_final.iloc[i[0]].title)
```

## Prueba

```python
recommend('Avatar')
```
